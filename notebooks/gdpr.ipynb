```python
import hashlib
import pandas as pd
```

```python
csv_path = "../data/raw/store-data-6aa6d7a3f171f140353680.csv"

df = pd.read_csv(csv_path)
```

```python

def hash_name(name):
    return hashlib.sha256(name.encode("utf-8")).hexdigest()

df["Customer Name"] = df["Customer Name"].apply(hash_name)
```
